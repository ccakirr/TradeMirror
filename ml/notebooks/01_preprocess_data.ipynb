## Data Cleaning

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../../data/raw/retail_trader_master.csv")

df.info()
```

```python
df.drop(["trader_id", "profitable", "net_return_pct"], axis=1, inplace=True)

len(df.columns)
```

```python
df.info()
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler

numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = df.select_dtypes(
    include=["object", "string"]
).columns

scaler = MinMaxScaler()

df_numeric = pd.DataFrame(
    scaler.fit_transform(df[numerical_columns]),
    columns=numerical_columns,
    index=df.index
)

encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

encoded_values = encoder.fit_transform(
    df[categorical_columns]
)

encoded_names = encoder.get_feature_names_out(
    categorical_columns
)

df_categorical = pd.DataFrame(
    encoded_values,
    columns=encoded_names,
    index=df.index
)

df_encoded = pd.concat(
    [df_numeric, df_categorical],
    axis=1
)

df_encoded.info()
```

```python
from pathlib import Path

output_path = Path(
    "/Users/ccakir/Desktop/TradeMiror/data/processed"
)

output_path.mkdir(parents=True, exist_ok=True)

df_encoded.to_csv(
    output_path / "encoded_data.csv",
    index=False
)
```
