## Data Cleaning & Feature Engineering

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../../data/raw/retail_trader_master.csv")

df.info()
```

```python
df.drop(["trader_id", "profitable", "net_return_pct"], axis=1, inplace=True)

len(df.columns)
```

```python
df["risk_amount"] = (
    df["starting_capital"] *
    df["risk_per_trade_pct"] / 100
)

df["leverage_risk"] = (
    df["avg_leverage"] *
    df["risk_per_trade_pct"]
)

df["experience_ratio"] = (
    df["prior_experience_years"] /
    df["age"].clip(lower=18)
)

df["trading_intensity"] = (
    df["trades_per_month"] /
    df["account_age_months"].clip(lower=1)
)

df["loss_win_hold_ratio"] = (
    df["avg_loss_hold_days"] /
    df["avg_win_hold_days"].clip(lower=0.1)
)

df["behavioral_risk"] = (
    df["revenge_trading"] +
    df["overconfidence"] +
    df["chases_volatility"]
)

df["discipline_score"] = (
    df["follows_plan"] +
    df["position_sizing_discipline"] +
    df["diversification"]
)
```

```python
df.info()
```

```python
from pathlib import Path

output_path = Path("../../data/processed")
output_path.mkdir(parents=True, exist_ok=True)


df.to_csv(output_path / "features.csv", index=False)

print("saved:", output_path / "features.csv", "| shape:", df.shape)
```
